```python
import plotly
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
df = pd.read_excel(r"C:\Users\changl3\OneDrive - Carrier Corporation\02_projects\13_meta\plots\c_SCF\1017_1830.xlsx", index_col = 0)
```

```python
# Update the plot table here to match your test file.
var_list = [

            [['D ECDW T', 'D LCDW T', 'D ECHW T', 'D LCHW T'],['D VFD SPD','D GV1 POS'], ['D SURG RGN']],
            [['U ECHW T', 'U LCHW T', 'U ECDW T', 'U LCDW T'], ['U VFD SPD','U GV1 POS'], ['U SURG RGN']],
            [['U KW PCT', 'D KW PCT'],['D HGBP POS', 'U HGBP POS'], ['D CHW FL']]
            
]
var_titles =  ['D LCDW/ECDW/ECW/LCW/', 'D GV/VFD', 'D surge_region',
               'U LCDW/ECDW/ECW/LCW/', 'U GV/VFD', 'U surge_region',
               'Power', 'HGBP', 'Flow Rate'
          ]
```

```python
row_num = len(var_list)
col_num = len(var_list[0])
fig = make_subplots(rows=row_num, cols=col_num, shared_xaxes='all', vertical_spacing=0.04, subplot_titles=var_titles)

for row_idx, row_list in enumerate(var_list):
    for col_idx, col_list in enumerate(row_list):
        for var in col_list:
            var = var
            fig.add_trace(go.Scatter(x=df.index, y=df[var], name=var), row=row_idx+1, col=col_idx+1)

fig.update_layout(height=800, width=1200)
fig.write_html('alc_1017_1830.html')
fig.show()
```
